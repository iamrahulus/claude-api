```python
from dotenv import load_dotenv
print('hello')
import os
#script_dir = os.path.dirname(os.path.abspath(__file__))
print(os.getcwd())
print(load_dotenv())
key = os.getenv('ANTHROPIC_API_KEY')
model = os.getenv('ANTHROPIC_MODEL')
from anthropic import Anthropic
client  = Anthropic()
print (client)
```

```python
message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=[
         {"role": "user", "content": "What is Quantum Computing? Answer in one sentence."}
    ]
)
```

```python
def extract_text(content_blocks):
    text_parts = []
    for block in content_blocks:
        if hasattr(block, "text"):
            text_parts.append(block.text)
    return "".join(text_parts)

extract_text(message.content)
```

```python
def add_user_message(messages, content):
    messages.append({"role": "user", "content": content})

def add_assistant_message(messages, content):
    messages.append({"role": "assistant", "content": content})

def chat(messages):
    response = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages
    )
    assistant_response = extract_text(response.content)
    return assistant_response



```

```python
messages = []
add_user_message(messages, "What is Quantum Computing? Answer in one sentence.")

answer = chat(messages)

add_assistant_message(messages, answer)


add_user_message(messages, "Write another sentence.")

print(messages)

answer = chat(messages)

answer
```

```python
messages = []
while True:
    user_input = input("> ")
    print('> ' + user_input)
    if user_input.lower() == "exit":
        break
    add_user_message(messages, user_input)
    answer = chat(messages)
    print('------')
    print('Assistant:', answer)
    print('------')
    add_assistant_message(messages, answer)
    
```
